## Different Delimiters

Some people just want to watch the world burn, so they create datasets where the columns are separted by something other than a comma.

#### Bad - Using another delimiter like `:`

##### `StarData_Ver2.dat`

```
StarID:Parallax:GMag
A5853:768.07:8.98
B4472:546.98:8.19
C3864:415.18:11.03
D7628:392.75:6.55
```

## Astropy

````python
star_table_2 = QTable.read('./Data/StarData_Ver2.dat', 
                           format='ascii.csv',
                           delimiter = ":")
````

````python
star_table_2
````

## Pandas

````python
p_star_table_2 = pd.read_csv('./Data/StarData_Ver2.dat', sep = ":")
````

````python
p_star_table_2
````

# Fixed-Width Data Tables

* These types of data tables are **VERY** common in astronomy
* The columns have a fixed-widths
* Whitespace is used to seperate columns **AND** used within columns

`StdStars.dat`

## Astropy

````python
my_column_start = [0, 12, 15, 18]
````

````python
my_column_names = ['Star', 'RAh', 'RAm', 'RAs']
````

````python
standard_table = QTable.read('./Data/StdStars.dat', 
                            format='ascii.fixed_width_no_header',
                            names = my_column_names,
                            col_starts = my_column_start
                            )
````

````python
standard_table
````

## Pandas

````python
my_colspecs = [(0, 11), (12, 14), (15, 17), (18, -1)]
````

````python
p_standard_table = pd.read_fwf('./Data/StdStars.dat', 
                               colspecs = my_colspecs,
                               names = my_column_names
                              )
````

````python
p_standard_table
````

### Pandas is actually really good at guessing the columns, so you can often just use:

````python
p_standard_table = pd.read_fwf('./Data/StdStars.dat', 
                               names = my_column_names
                              )
````

````python
p_standard_table
````

# VO Tables

Virtual Observatory (VO) tables are a new format developed by the International Virtual Observatory Alliance. 

It is a format based on the Extensible Markup Language (XML).

VO Tables are nice because they preserve the `Units` of a Astropy `QTable`

- Since a VO table has units, it does not make sense to use pandas
